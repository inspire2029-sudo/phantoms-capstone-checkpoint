# Logical Bugs Code Review

### **الكود المتصحح**

```python
def process_team_scores(scores_list):
    """
    بتحسب متوسط الدرجات وبتطلع الدرجات الناجحة
    والدرجة الناجحة هي 50 أو أكتر
    """

    try:
        if not scores_list:
            raise ValueError("قائمة الدرجات مينفعش تكون فاضية")

        total_score = 0
        passed_members = []

        # بنلف على كل الدرجات الموجودة في القائمة
        for score in scores_list:
            total_score += score

            # الدرجة 50 أو أكتر تعتبر ناجحة
            if score >= 50:
                passed_members.append(score)

        average = total_score / len(scores_list)

        return average, passed_members

    except TypeError:
        raise TypeError("لازم الدرجات تكون قائمة من الأرقام")
```

### **اختبار 1 — درجات عادية**

```python
sample_scores = [45, 50, 75, 90, 60]

avg, passed = process_team_scores(sample_scores)

print("المتوسط:", avg)
print("الدرجات الناجحة:", passed)
```

### **اختبار 2 — اختبار الدرجة الحدية**

```python
scores = [49, 50, 51]

avg, passed = process_team_scores(scores)

print("المتوسط:", avg)
print("الدرجات الناجحة:", passed)
```

### **اختبار 3 — مفيش درجات ناجحة**

```python
scores = [20, 30, 40]

avg, passed = process_team_scores(scores)

print("المتوسط:", avg)
print("الدرجات الناجحة:", passed)
```

### **اختبار 4 — كل الدرجات ناجحة**

```python
scores = [50, 60, 70]

avg, passed = process_team_scores(scores)

print("المتوسط:", avg)
print("الدرجات الناجحة:", passed)
```

### **اختبار 5 — قايمة درجات فاضية**

```python
try:
    process_team_scores([])
except ValueError as error:
    print("تم التعامل مع الخطأ:", error)
```

### **اختبار 6 — إدخال مش صح**

```python
try:
    process_team_scores("بيانات غلط")
except TypeError as error:
    print("تم التعامل مع الخطأ:", error)
```

### **اختبار نهائي للتأكد من النتائج**

```python
avg, passed = process_team_scores([45, 50, 75, 90, 60])

assert avg == 64.0
assert passed == [50, 75, 90, 60]

avg, passed = process_team_scores([49, 50, 51])

assert avg == 50.0
assert passed == [50, 51]

avg, passed = process_team_scores([20, 30, 40])

assert avg == 30.0
assert passed == []

avg, passed = process_team_scores([50, 60, 70])

assert avg == 60.0
assert passed == [50, 60, 70]

print("كل الاختبارات نجحت")
```

# تقرير مراجعة وتصحيح الأخطاء المنطقية

## الخطأ الأول — آخر درجة في القايمة مش بتتحسب

في الكود الأصلي كان الـ loop مكتوب بالشكل ده

`range(len(scores_list) - 1)`

وده معناه إن البرنامج بيلف على الدرجات لحد قبل آخر عنصر في القايمة

فلو عندنا الدرجات

`[45, 50, 75, 90, 60]`

البرنامج كان بيتجاهل الدرجة 60

وبالتالي الدرجة دي مكانتش بتدخل في حساب المجموع ومكانش بيتم فحصها لمعرفة إذا كانت ناجحة ولا لأ

وده بيأثر على المتوسط وعلى قايمة الدرجات الناجحة

و حل المشكلة تم باستخدام

`for score in scores_list`

وبكده البرنامج بيلف على كل الدرجات الموجودة في القايمة

## الخطأ الثاني — الدرجة 50 مش بتتحسب ناجحة

الكود الأصلي كان بيستخدم الشرط

`if score > 50`

وده معناه إن الدرجة لازم تكون أكبر من 50

لكن المطلوب إن الدرجة الناجحة تكون 50 أو أكتر

فبالتالي الدرجة 50 كانت بتتحسب رسوب بالغلط

تم تصحيح الشرط لـ

`if score >= 50`

وبكده الدرجة 50 أصبحت ناجحة زي ما هو مطلوب

## الخطأ الثالث — البرنامج ممكن يقع بسبب القسمة على صفر

الكود الأصلي كان بيحسب المتوسط باستخدام

`total_score / len(scores_list)`

ولو المستخدم دخل قايمة فاضية هيكون عدد الدرجات صفر

وساعتها البرنامج هيحاول يقسم على صفر وده هيؤدي إلى حدوث `ZeroDivisionError`

تم التعامل مع الحالة دي عن طريق التأكد الأول إن القايمة مش فاضية

ولو كانت فاضية بيتم رفع `ValueError` برسالة واضحة بدل ما البرنامج يقع بشكل غير مفهوم

## معالجة الاستثناءات

تم إضافة `try-except` للتعامل مع الإدخالات الـ غلط

وكمان تم اختبار حالة القايمة الفاضية وحالة إدخال بيانات غير مناسبة
