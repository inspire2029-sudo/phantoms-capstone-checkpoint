**`Dictionaryال`**

```python
students = {
    "Hana": 90,
    "Ahmed": 85,
    "Menna": "Absent",
    "Ali": 70
}
```

**` Function ال `**

```python
def calculate_average_score(students):
    scores = []

    try:
        for score in students.values():
            try:
                scores.append(float(score))
            except (ValueError, TypeError):
                continue

        if not scores:
            return 0

        return sum(scores) / len(scores)

    except (TypeError, AttributeError):
        return 0
```

**` Function استدعاء ال`**

```python
average_score = calculate_average_score(students)

print(f"Average score: {average_score:.2f}")
```

**` الكود كامل`**

```python
students = {
    "Hana": 90,
    "Ahmed": 85,
    "Menna": "Absent",
    "Ali": 70
}


def calculate_average_score(students):
    scores = []

    try:
        for score in students.values():
            try:
                scores.append(float(score))
            except (ValueError, TypeError):
                continue

        if not scores:
            return 0

        return sum(scores) / len(scores)

    except (TypeError, AttributeError):
        return 0


average_score = calculate_average_score(students)

print(f"Average score: {average_score:.2f}")
```

**`اختبار لو القايمه فارغة`**

```python
empty_students = {
    "Hana": "Absent",
    "Ahmed": "Absent"
}

print(calculate_average_score(empty_students))
```
