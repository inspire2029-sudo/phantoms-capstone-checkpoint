# استدعاء المكتبه

```python
import sqlite3
```

## انشاء قاعدة بيانات والاتصال بيها

```python
conn = sqlite3.connect("store.db")
```

## انشاء كوسور

```python
cursor = conn.cursor()
```

## انشاء جدول المنتجات

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY,
    name TEXT,
    category TEXT,
    price REAL
)
""")
```

## حفظ التغيير

```python
conn.commit()
```

## ادخال المنتجات

```python
products = [
    (1, "Laptop", "Electronics", 25000),
    (2, "Mouse", "Electronics", 800),
    (3, "Desk", "Furniture", 3500),
    (4, "Headphones", "Electronics", 2200),
    (5, "Chair", "Furniture", 4500)
]

cursor.executemany("""
INSERT OR IGNORE INTO products (id, name, category, price)
VALUES (?, ?, ?, ?)
""", products)

conn.commit()
```

##  الاستعلام الأول WHERE

```python
cursor.execute("""
SELECT * FROM products
WHERE category = 'Electronics'
""")

results = cursor.fetchall()

print("Products in Electronics category:")

for product in results:
    print(product)
```

## الاستعلام الثاني ORDER BY

```python
cursor.execute("""
SELECT * FROM products
ORDER BY price DESC
""")

results = cursor.fetchall()

print("Products sorted by price from highest to lowest:")

for product in results:
    print(product)
```

## قفل الاتصال

```python
conn.close()
```

# الكود كامل

```python
import sqlite3

conn = sqlite3.connect("store.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY,
    name TEXT,
    category TEXT,
    price REAL
)
""")

conn.commit()

products = [
    (1, "Laptop", "Electronics", 25000),
    (2, "Mouse", "Electronics", 800),
    (3, "Desk", "Furniture", 3500),
    (4, "Headphones", "Electronics", 2200),
    (5, "Chair", "Furniture", 4500)
]

cursor.executemany("""
INSERT OR IGNORE INTO products (id, name, category, price)
VALUES (?, ?, ?, ?)
""", products)

conn.commit()

cursor.execute("""
SELECT * FROM products
WHERE category = 'Electronics'
""")

results = cursor.fetchall()

print("Products in Electronics category:")

for product in results:
    print(product)

cursor.execute("""
SELECT * FROM products
ORDER BY price DESC
""")

results = cursor.fetchall()

print("\nProducts sorted by price from highest to lowest:")

for product in results:
    print(product)

conn.close()
```
