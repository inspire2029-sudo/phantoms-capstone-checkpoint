## 1) NumPy Practice

هنستخدم NumPy لإنشاء Array من الأرقام
وبعد كده نحسب الـ Mean والـ Standard Deviation

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Numbers:", numbers)
```

```python
mean = np.mean(numbers)

print("Mean:", mean)
```

```python
std = np.std(numbers)

print("Standard Deviation:", std)
```

## بقصر طول الـ اوت بوت

```python
print(f"Mean: {mean:.2f}")
print(f"Standard Deviation: {std:.2f}")
```
