## 3) Handling Missing Values

مش كل Missing Values هنتعامل معاها بنفس الطريقة

القرار بيعتمد على نوع البيانات ونسبة القيم الناقصة وأهمية العمود

- age: هنستخدم Median لأن العمود رقمي وفيه عدد كبير من القيم الناقصة والـ Median أكثر مقاومة للقيم المتطرفة
- embarked: هنستخدم Mode لأن العمود Categorical وعدد القيم الناقصة قليل جدًا
- embark_town: هنحذف العمود لأنه يحتوي على نفس معلومات embarked بشكل مكرر
- deck: هنحذف العمود لأن نسبة القيم الناقصة فيه كبيرة جدًا وبالتالي تعويضها هيكون غير موثوق

```python
df["family_size"] = df["sibsp"] + df["parch"] + 1

df[["sibsp", "parch", "family_size"]].head()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["age"].isnull().sum()
```

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

```python
df["embarked"].isnull().sum()
```

```python
df = df.drop(columns=["deck", "embark_town", "class", "who", "adult_male", "alive"])
```

```python
df.info()
```

```python
df["sex"].value_counts()
```

```python
df.groupby("pclass")["survived"].mean()
```

```python
features = [
    "pclass",
    "sex",
    "age",
    "sibsp",
    "parch",
    "fare",
    "embarked",
    "alone",
    "family_size"
]

label = "survived"

X = df[features]
y = df[label]

print("Features:")
print(list(X.columns))

print("\nLabel:")
print(y.name)
```

```python
X = pd.get_dummies(
    X,
    columns=["sex", "embarked"],
    drop_first=True,
    dtype=int
)

X.head()
```

```python
X.info()
```

```python
cleaned_df = X.copy()
cleaned_df[label] = y

cleaned_df.head()
```

```python
print("Missing Values:")
print(cleaned_df.isnull().sum())

print("\nShape:")
print(cleaned_df.shape)

print("\nData Types:")
print(cleaned_df.dtypes)
```

```python
cleaned_df.to_csv("cleaned_data.csv", index=False)
```
