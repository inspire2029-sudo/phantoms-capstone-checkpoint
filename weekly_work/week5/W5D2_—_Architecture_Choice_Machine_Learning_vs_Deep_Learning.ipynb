# PHANTOMS | W5D2
## Architecture Choice: Machine Learning vs. Deep Learning

### Engineering Decision Report

**Task:** Choosing the appropriate model architecture for apartment price prediction

</div>

<div dir="rtl" align="right">

## 1. نظرة عامة

المطلوب هو اختيار التقنية المناسبة لبناء نموذج يتوقع سعر شقة اعتمادًا على 10 خصائص رقمية فقط، بدون تحديد تقنية معينة مسبقًا.

المشكلة هنا مش إننا نستخدم الـ Deep Learning لمجرد إنه تقنية أكثر تقدمًا، لكن المطلوب هو اتخاذ قرار هندسي مبني على طبيعة البيانات، حجمها، تعقيد المشكلة، وتكلفة الحل.

</div>

<div dir="rtl" align="right">

## 2. تعريف المشكلة

المشكلة تعتبر **Regression Problem (مشكلة انحدار)** لأن الهدف هو التنبؤ بقيمة رقمية مستمرة، وهي سعر الشقة.

ممكن تكون الـ 10 Features مثل:

- مساحة الشقة
- عدد الغرف
- عدد الحمامات
- الدور
- عدد أدوار المبنى
- عمر المبنى
- المسافة من وسط المدينة
- مستوى التشطيب
- وجود موقف سيارات
- الموقع

الـ Target Variable في الحالة دي هو **Apartment Price**.

</div>

<div dir="rtl" align="right">

## 3. طبيعة البيانات

البيانات المستخدمة هنا هي **Tabular Data (بيانات جدولية)**.

كل شقة تمثل Row، وكل خاصية من خصائص الشقة تمثل Column.

مثال:

</div>

| Area | Rooms | Bathrooms | Floor | Age | Distance | Price |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| 120 | 3 | 2 | 4 | 8 | 5.2 km | 2,500,000 |
| 150 | 4 | 2 | 7 | 3 | 3.1 km | 3,200,000 |

<div dir="rtl" align="right">

وده مختلف عن **Unstructured Data (بيانات غير منظمة)** زي الصور والصوت والفيديو والنصوص.

الـ Deep Learning بيكون مفيد جدًا في أنواع البيانات دي لأنه يقدر يتعلم Features وتمثيلات معقدة منها بشكل تلقائي.

لكن في حالتنا، الـ Features محددة بالفعل وفي صورة رقمية وجدولية، وبالتالي مفيش احتياج واضح لاستخدام شبكة عصبية عميقة لاستخراج Features معقدة.

</div>

<div dir="rtl" align="right">

## 4. حجم البيانات وتأثيره على الاختيار

لو افترضنا إن Dataset عبارة عن عدد صغير أو متوسط من سجلات الشقق، فاختيار Deep Learning مش هيكون مبرر بشكل قوي.

الشبكات العصبية العميقة تحتوي على عدد كبير من **Parameters (المعاملات)** تحتاج إلى بيانات كافية للتعلم والتعميم.

مع Dataset صغيرة، بيزيد خطر حدوث **Overfitting (فرط التوافق)**، يعني إن النموذج يتعلم تفاصيل بيانات التدريب بشكل زائد بدل ما يتعلم العلاقة العامة بين خصائص الشقة والسعر.

وبالتالي ممكن يحقق النموذج نتائج ممتازة على Training Data، لكن أداؤه يقل عند التعامل مع بيانات جديدة.

## 6. تحليل النماذج التقليدية

### 6.1 Linear Regression

ممكن نستخدم **Linear Regression** كـ **Baseline Model (نموذج مرجعي)** لأنه بسيط وسريع وسهل نسبيًا في تفسير العلاقة بين الـ Features والسعر.

لكن المشكلة إن العلاقة بين خصائص العقار والسعر مش لازم تكون خطية بالكامل.

### 6.2 Random Forest

**Random Forest** مناسب جدًا للـ Tabular Data، ويقدر يتعامل مع العلاقات غير الخطية بين الـ Features والـ Target.

كمان بيحتاج إعداد أقل تعقيدًا من بناء شبكة Deep Learning كاملة.

### 6.3 XGBoost

**XGBoost** يعتبر اختيار قوي لمشاكل الـ Tabular Regression، وبيقدر يتعامل مع العلاقات المعقدة بين الخصائص المختلفة.

عشان كده ممكن يكون واحد من النماذج الأساسية اللي نجربها بعد الـ Baseline.

</div>

## 7. سرعة التدريب والتعقيد

في Dataset صغيرة أو متوسطة تحتوي على 10 Features فقط، النماذج التقليدية غالبًا هتكون أسرع وأسهل في التدريب.

استخدام Deep Learning ممكن يتطلب تحديد عدد الـ Layers، وعدد الـ Neurons، والـ Learning Rate، والـ Batch Size، وعدد الـ Epochs، بالإضافة إلى عملية **Hyperparameter Tuning**.

ده بيضيف مستوى من التعقيد مش ضروري طالما إن المشكلة ممكن تتحل بنموذج أبسط.

</div>

<div dir="rtl" align="right">

## 8. Interpretability — قابلية تفسير النموذج

تفسير النتائج مهم في مشكلة زي أسعار العقارات، لأننا محتاجين نفهم إيه العوامل اللي النموذج بيعتمد عليها في التنبؤ.

ممكن نستخدم النماذج التقليدية وأدوات التفسير لمعرفة تأثير Features مثل:

- مساحة الشقة
- الموقع
- عدد الغرف
- عمر المبنى
- الدور

أما الـ Deep Neural Networks فعادةً بتكون أكثر تعقيدًا في تفسير سبب الوصول إلى Prediction معين.

</div>

<div dir="rtl" align="right">

## 9. Computational Cost — التكلفة الحسابية

في الحالة الحالية عندنا:

**10 Features + Tabular Data + Dataset صغيرة أو متوسطة**

وبالتالي مش محتاجين بنية Deep Learning كبيرة أو موارد حسابية مرتفعة لمجرد تنفيذ المهمة.

استخدام Traditional Machine Learning هيكون أبسط من ناحية:

- وقت التدريب
- استهلاك الذاكرة
- الموارد الحسابية
- إعداد النموذج
- تجربة أكثر من Model
- الصيانة المستقبلية

</div>

<div dir="rtl" align="right">

## 10. Overfitting Risk

واحد من أهم الأسباب اللي تخلي استخدام Deep Learning غير ضروري هنا هو خطر الـ **Overfitting**.

لو استخدمنا Neural Network كبيرة على Dataset صغيرة، ممكن يكون **Model Complexity** أعلى من كمية المعلومات المتاحة للتعلم.

في الحالة دي النموذج ممكن يحفظ تفاصيل Training Data بدل ما يتعلم Patterns عامة تقدر تساعده في التنبؤ ببيانات جديدة.

عشان كده لازم يكون حجم وتعقيد النموذج متناسب مع حجم وطبيعة البيانات.

</div>

### Recommended Starting Approach

**Linear Regression → Baseline**

**Random Forest / XGBoost → Advanced Models**

القرار هنا هو البدء بـ **Traditional Machine Learning** بدل Deep Learning.

السبب الأساسي إن المشكلة تعتمد على عدد محدود من الـ Numerical Features والبيانات في صورة Tabular Data.

هنبدأ بنموذج بسيط كـ Baseline، وبعدها نقارن نماذج أكثر قوة مثل Random Forest أو XGBoost.

## 11. Model Evaluation

بعد اختيار النماذج، ممكن نقارن أداءها باستخدام Metrics مناسبة لمشكلة الـ Regression، مثل:

- **MAE — Mean Absolute Error**
- **RMSE — Root Mean Squared Error**
- **R² — R-squared**

المقارنة الفعلية بين النماذج لازم تعتمد على نتائج الاختبار، وليس فقط على اسم أو تعقيد الـ Algorithm.

</div>

<div dir="rtl" align="right">

## 12. إمتى ممكن نستخدم Deep Learning؟

القرار ممكن يتغير لو طبيعة المشكلة أو حجم البيانات اتغير.

مثلاً لو أصبح عندنا:

- صور للشقق
- وصف نصي للعقارات
- فيديوهات
- صور للمناطق
- Dataset ضخمة جدًا
- ملايين السجلات

في الحالة دي ممكن يكون للـ Deep Learning دور أكبر.

مثلاً ممكن استخدام **Computer Vision** لتحليل الصور، أو **NLP / Transformers** لتحليل النصوص.

لكن بالنسبة للمشكلة الحالية، مفيش احتياج واضح للمستوى ده من التعقيد.

</div>

<div dir="rtl" align="right">

## 13. الخلاصة

الاختيار بين Machine Learning وDeep Learning لازم يعتمد على طبيعة المشكلة والبيانات، مش على كون تقنية معينة أحدث أو أكثر تعقيدًا.

في مشكلة توقع أسعار الشقق باستخدام 10 خصائص رقمية، **Traditional Machine Learning** يعتبر نقطة بداية مناسبة لأنه يتعامل بشكل جيد مع Tabular Data، ويحتاج موارد وتعقيد أقل، ويوفر خيارات جيدة للتفسير والمقارنة.

وفي نفس الوقت، القرار مش ثابت للأبد. لو حجم البيانات أو طبيعتها اتغيرت، ممكن نعيد تقييم الـ Architecture ونشوف إذا كان استخدام Deep Learning أصبح له مبرر فعلي.
